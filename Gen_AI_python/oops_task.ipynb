2. Write a class Student with attributes name and marks. Add a method display() to print details.

3. Create a class Rectangle with attributes length and width. Add methods to calculate area and perimeter.

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

obj = Rectangle(10, 5)

print("Area:", obj.area())
print("Perimeter:", obj.perimeter())
```

5. Define a class Circle with a class variable pi = 3.14. Add a method to calculate the area.

```python
class Circle:
    pi = 3.14

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return Circle.pi * self.radius * self.radius

obj = Circle(5)

print("Area of Circle:", obj.area())
```

6. Create a class BankAccount with attributes account_number, name, and balance. Add methods to deposit, withdraw, and display balance.

```python
class BankAccount:
    def __init__(self, account_number, name, balance):
        self.account_number = account_number
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    def display_balance(self):
        print("Account Number:", self.account_number)
        print("Name:", self.name)
        print("Balance:", self.balance)


obj = BankAccount(101, "Pravina", 5000)

obj.deposit(2000)
obj.withdraw(1000)
obj.display_balance()
```

7. Write a class Employee with attributes emp_id, name, and salary. Add a method to give a 10% bonus to employees.

```python
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    def bonus(self):
        self.salary += self.salary * 0.10

    def display(self):
        print("Employee ID:", self.emp_id)
        print("Name:", self.name)
        print("Salary:", self.salary)


obj = Employee(101, "Pravina", 100000)

obj.bonus()
obj.display()
```

8. Implement a class Library with methods to add_book(title), remove_book(title), and display_books().

```python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title):
        self.books.append(title)
        print(title, "added successfully")

    def remove_book(self, title):
        if title in self.books:
            self.books.remove(title)
            print(title, "removed successfully")
        else:
            print("Book not found")

    def display_books(self):
        print("Available Books:", self.books)


obj = Library()

obj.add_book("Python")
obj.add_book("Java")
obj.add_book("SQL")

obj.display_books()

obj.remove_book("Java")

obj.display_books()
```

9. Create a class Calculator with methods add, subtract, multiply, and divide. Demonstrate with objects.

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divide(self, a, b):
        return a / b


obj = Calculator()

print("Addition:", obj.add(10, 5))
print("Subtraction:", obj.subtract(10, 5))
print("Multiplication:", obj.multiply(10, 5))
print("Division:", obj.divide(10, 5))
```

10. Write a program to show the difference between instance variables and class variables with an example.



    Class variables:
     - Variables declared inside a class but outside methods.
     - Shared by all objects of the class.

     
    Instance variables:
     - Variables declared using self inside methods.
     - Each object has its own copy.

```python
class Student:
    college = "JNN College"  # Class variable

    def __init__(self, name, marks):
        self.name = name       # Instance variable
        self.marks = marks     # Instance variable


obj1 = Student("Pravina", 95)
obj2 = Student("divya", 85)

print("Student 1:", obj1.name, obj1.marks, obj1.college)
print("Student 2:", obj2.name, obj2.marks, obj2.college)
```
