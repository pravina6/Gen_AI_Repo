```python
"""-----
    ----
     ---
      --
       -"""


row = 5

for i in range(row):
    for j in range(i):
        print(" ", end="")

    for j in range(row - i):
        print("-", end="")

    print()
```

```python
""" *********
     *******
      *****
       ***
        *   """

num = int(input("Enter a number: "))

for i in range(num, 0, -1):
    for j in range(num - i):
        print(" ", end="")

    for k in range(2 * i - 1):
        print("*", end="")

    print()
```
