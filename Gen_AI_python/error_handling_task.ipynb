1. What is the difference between syntax errors and runtime errors in Python?


Syntax Error:
- Occurs when we write code that does not follow Python's grammar rules.
- Python cannot execute the code until the syntax error is corrected.

```python
if 10 > 5
    print("True")
```

Runtime Error:
- Occurs while the program is running.
- The syntax is correct, but an error occurs during execution.

Difference: 
Syntax errors occur before execution, whereas runtime errors occur during execution.

```python
a = 10
b = 0
print(a / b)
```

2. Explain the role of the try, except, else, and finally blocks.


  These blocks are used for handling exceptions in Python.
- try: Contains the code that may cause an exception.
- except: Handles the exception if an error occurs.
- else: Executes only if no exception occurs in the try block.
- finally: Executes whether an exception occurs or not.

```python
try:
    a = 10
    b = 2
    result = a / b

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Result:", result)

finally:
    print("Program completed")
```

3. What happens if an exception is raised but not handled

  If an exception is not handled:
- The program stops executing at that point.
- Python displays an error message called a traceback.
- The remaining code in that execution path will not execute.

```python
a = 10
b = 0

print(a / b)

print("Hello")
```

4. What is the difference between Exception and BaseException in Python

      Both are built-in exception classes in Python.


- BaseException: 
     The base class for all built-in exceptions.
     BaseException is the parent class

       Examples of direct BaseException subclasses:
         - KeyboardInterrupt
         - SystemExit
         - GeneratorExit


- Exception: 
       A subclass of BaseException that includes most regular application errors.
       Exception is its subclass.

       Examples of Exception:
          - ValueError
          - TypeError
          - ZeroDivisionError
          - IndexError

5. Explain the use of the raise keyword with an example


      The raise keyword is used to manually generate an exception in Python.

       Explanation:
        - If the age is less than 18, the raise keyword generates a ValueError.
        - The program stops unless the exception is handled.

```python
age = 15

if age < 18:
    raise ValueError("Age must be 18 or above")

print("Eligible")
```

6. How can you create a custom exception class in Python?
    
    We can create our own exception by defining a class that inherits from the built-in Exception class.


      Explanation:
        - MyError is our custom exception class.
        - It inherits from the Exception class.
        - raise is used to generate our custom exception.

```python
class MyError(Exception):
    pass

try:
    raise MyError("This is a custom exception")

except MyError as e:
    print(e)
```

7. What is the purpose of the assert statement, and how does it differ from raising an exception?


        Assert statement:

         The assert statement is used to check whether a condition is True.

         If the condition is False, Python raises an AssertionError.


        Difference:
            Assert	                                        Raise
            Used to check conditions.	                        Used to manually generate exceptions.
            Raises AssertionError if condition is False.	Can raise different exception types.
            Mainly used for debugging.	             Used for application error handling and validation.
            Can be disabled with Python optimization.	        Cannot be disabled in the same way.

```python
age = 15

assert age >= 18, "Age must be 18 or above"

print("Eligible")
```

8. Can we use multiple except blocks in Python? Give an example.


     Yes. We can use multiple except blocks to handle different types of exceptions separately.



       Explanation:
        - ValueError handles invalid input.
        - ZeroDivisionError handles division by zero.
        - Exception handles other regular exceptions.
        Python executes the matching except block.

```python
try:
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))

    print(a / b)

except ValueError:
    print("Please enter numbers only")

except ZeroDivisionError:
    print("Cannot divide by zero")

except Exception:
    print("Something went wrong")
```

9. What happens if you use a finally block without an except block?


      Yes, we can use a finally block without an except block.

      The finally block executes whether an exception occurs or not.


       Explanation:
        - The try block generates an exception.
        - The finally block executes.
        - The exception is then propagated because it was not handled.

```python
try:
    print("Inside try block")
    a = 10 / 0

finally:
    print("Finally block executed")
```

10. What is the difference between checked and unchecked exceptions? Does Python support checked exceptions?

        Checked Exceptions:
         - These exceptions must be handled or declared by the programmer.
         - They are checked at compile time.
         - Example: Java has checked exceptions such as IOException.


        Unchecked Exceptions:
         - These exceptions do not need to be explicitly handled or declared.
         - They generally occur during runtime.
         - Example: Java's NullPointerException and ArithmeticException.
