35.Write a program using while loop to check whether a number is a palindrome.

```python
num =  101

orginal = num 
reversed = 0
while  num>0:
  digit = num%10
  reversed = reversed * 10 + digit
  num = num // 10

if orginal == reversed:
    print("palindrome")
else:
    print("not palindrome")
```
