### Python Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.figure() # give the figure o/p : <Figure size 640x480 with 0 Axes>
plt.plot() # to show the image use plot() o/p :[] 
plt.show(); # to remove the square bracket use show();
```

### Scatter Plot 

```python
x=np.array([1,2,3,4])
y=np.array([1,2,3,4])

plt.scatter(x,y,c='crimson',marker="p",s=100) # here 1st sholud be hori 2nd vertical , 
                                         #to change color we can use c or color inside scatter ()  
                                        # to change the shape we can use marker ()
                                        # to change sixe we can use s="value"
plt.show(); # scatter dot graph
            # default color of graph is blue 
            # (1,2,3)-major ticks 
            # (1.0,1.2,1.3)- like that points are minor ticks
```

```python
x=np.array([1,2,3,4])
y=np.array([1,2,3,4])
color=["maroon","cyan","peru","seagreen"]
plt.scatter(x,y,c=color,marker='p',s=100) # scatter can accept only one marker at a time or we want we can used in the for loop 
plt.show();
```

### Line graph 

```python
x=np.array([2,1,3,4])
y=np.array([1,2,4,3])
color=["maroon","cyan","peru","hotpink"]
plt.plot(x,y,c="g",linestyle="dashdot") # line style will change the line 
plt.scatter(x,y,c=color,marker='P',s=100) # if want dot we can use scatter 
plt.show();
```

### Bar chart 

```python
year=["1990","2000","2010","2020","2030"]
lang=["none","Java","C++","javascript","Python"] # in string 1st value in bar can't visible
# lang=[1,2,3,4,5] # but in numberical value visible
c=["green","crimson","darkorange","forestgreen","lightpink"]
plt.bar(year,lang,color = c , align="edge")  # in scatter , line for color we can use c but in bar we have to use color
plt.show();

    
```

### Histogram (frequency distribution)

```python
np.random.seed(2)
a=np.random.normal(50,2,100) # 50 is average 50 will be center , 2 is standard deviation , 100 random value
plt.hist(a , color='crimson', edgecolor="black",bins=20) # bins is that bar 
plt.show();
```

### Pie chart 

```python
lang = ["Python","Java","C++","C"]
votes= [90,75,50,40]
c=["lightcoral","tomato","steelblue","plum"]
ex=[0.1,0,0.2,0]
plt.pie(votes , labels=lang, colors=c , explode=ex) # in pie have to use colors , explode is seperate value 
plt.show();
```

```python
years=["2000","2010","2020","2030"]
income = [200,400,600,800]
plt.title("daily income")
plt.xlabel("years")
plt.ylabel("income")
plt.plot(years,income,c="navy",marker="*")
plt.show();
```

### Lengend()

```python
stocka=[1,2,3,4,5]
stockb=[3,4,5,1,2]
stockc=[5,3,4,2,1]
plt.title("STOCK MARKET")
plt.plot(stocka, label="stock A",marker="o")
plt.plot(stockb, label="stock B",marker="p")
plt.plot(stockc, label="stock C",marker="*")
plt.legend()
plt.show();
```

```python
### plt from dict 

car_milage={
    "BMW":10,
    "ford":7,
    "toyota":8
}
fig,ax=plt.subplots()
ax.bar(car_milage.keys(),car_milage.values())
plt.show();
```

```python
### to convert into horizontal graph 

### * add h to graph like ax.barh , ax.scatterh , ax.ploth like that 
### * convert 1st value into list 

car_milage={
    "BMW":10,
    "ford":7,
    "toyota":8
}
fig,ax=plt.subplots()
ax.barh(list(car_milage.keys()),car_milage.values())
plt.show();
```

### subplots comparing age vs cholestral age vs thalach 

```python
fig,ax=plt.subplots(nrows=2,ncols=1,figsize=(12,6))
#subplot 1

scatter=ax[0].scatter(df["age"],df["chol"],c=df["target"],cmap="Accent")
#Customise plot
ax[0].set(title="AGE VS CHOLESTRAL",xlabel="Age",ylabel="Cholestral")
ax[0].legend(*scatter.legend_elements(),title="target")
#add mean value 
ax[0].axhline(df["chol"].mean(),c='r',linestyle="dashed")


#subplot 2

scatter=ax[1].scatter(df["age"],df["thalach"],c=df["target"],cmap="magma")
ax[1].set(title="AGE VS THALACH",xlabel="Age",ylabel="Thalach")
ax[1].legend(*scatter.legend_elements(),title="target")
ax[1].axhline(df["thalach"].mean(),c="r",linestyle="dashed")

plt.subplots_adjust(hspace=(0.4))
plt.savefig("comparing age vs cholestral and age vs thalach ")
plt.show();
```

### Image savefig 

```python
fig,ax=plt.subplots(nrows=1,ncols=1,figsize=(10,6))
scatter=ax.scatter(df["age"],df["chol"],c=df["target"],cmap="winter")
ax.set(title="heart disease and cholestral level",xlabel="age",ylabel="chol")
ax.legend(*scatter.legend_elements(),title="target")
ax.axhline(df["chol"].mean(),c="r",linestyle="dashed")
plt.savefig("probability of heart disease ")
plt.show();
```
