if , elif , else rules:

* if should have only one if in the block (mandatory)
* elif can have many times has much you want it (optional)
* else should have only one else in the block and no condition (optional)

SIMPLE GUESS THE NUMBER PROGRAM

```python
guess=int(input("enter the number :"))
score=100
secret_number=26
if guess==secret_number:
    print("your guess is correct ")
    score=score+20
    print("your score is ", score)
elif guess > secret_number:
    print("your guess is high")
    score=score-10
    print("your score is ", score)
elif guess < secret_number and guess > 0:
    print("your guess is low ")
    score=score-10
    print("your score is ", score)
else:
    print("guess is invalid ")
```
